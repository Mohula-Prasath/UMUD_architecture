```python
import os
import io
import math
import time
import warnings
from pathlib import Path
from collections import defaultdict

warnings.filterwarnings("ignore")
os.environ["PYTHONWARNINGS"] = "ignore"
os.environ["OPENCV_LOG_LEVEL"] = "SILENT"

import cv2
import numpy as np
import pandas as pd
from PIL import Image
from scipy.signal import savgol_filter
np.seterr(all="ignore")

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torch.cuda.amp import autocast, GradScaler

RNG_SEED = 42
np.random.seed(RNG_SEED)
torch.manual_seed(RNG_SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(RNG_SEED)

DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")
device_label = torch.cuda.get_device_name(0) if torch.cuda.is_available() else "CPU Execution Mode"

print("+----------------------------------------------------------------+")
print("| CONTINUUM BIOMECHANICAL MORPHOMETRY RUNTIME INITIALIZED        |")
print("+----------------------------------------------------------------+")
print(f"| Compute Hardware Platform : {DEV.type.upper():<34} |")
print(f"| Physical Accelerator Model: {device_label:<34} |")
print(f"| Mixed Precision (AMP)     : Enabled (FP16 Autocast)            |")
print(f"| Deterministic Random Seed : {RNG_SEED:<34} |")
print("+----------------------------------------------------------------+")
```

```python
SEARCH_ROOTS = [
    Path("/kaggle/input/umud-challenge-muscle-architecture-in-ultrasound-data"),
    Path("/kaggle/input/competitions/umud-challenge-muscle-architecture-in-ultrasound-data"),
    Path("data/raw"),
]

DATASET_DIR = None
for cand in SEARCH_ROOTS:
    if cand.is_dir():
        DATASET_DIR = cand
        break

if DATASET_DIR is None:
    raise FileNotFoundError("UMUD Challenge dataset mount point not found.")

def list_image_files(folder):
    if folder is None or not Path(folder).exists():
        return []
    valid_exts = {".tif", ".tiff", ".png", ".jpg", ".jpeg"}
    return sorted([
        p for p in Path(folder).rglob("*")
        if p.is_file() and p.suffix.lower() in valid_exts and not p.name.startswith(".")
    ])

img_dir_options, mask_dir_options = [], []
eval_dir_options = []

for root, dirs, _ in os.walk(DATASET_DIR):
    for d in dirs:
        p = Path(root) / d
        name = str(p).lower()
        if "apo" in name and ("img" in name or "image" in name) and "mask" not in name:
            img_dir_options.append(p)
        elif "apo" in name and "mask" in name:
            mask_dir_options.append(p)
        elif "test" in name:
            eval_dir_options.append(p)

def largest_folder(candidates):
    top_folder, top_count = None, 0
    for cand in candidates:
        n_found = len(list_image_files(cand))
        if n_found > top_count:
            top_count = n_found
            top_folder = cand
    return top_folder

IMG_FOLDER   = largest_folder(img_dir_options)
MASK_FOLDER  = largest_folder(mask_dir_options)
EVAL_FOLDER      = largest_folder(eval_dir_options)

all_scans = list_image_files(IMG_FOLDER)
all_masks  = list_image_files(MASK_FOLDER)
eval_files = list_image_files(EVAL_FOLDER)

stem_to_mask = {p.stem: p for p in all_masks}
paired_imgs, paired_masks = [], []
for p in all_scans:
    if p.stem in stem_to_mask:
        paired_imgs.append(p)
        paired_masks.append(stem_to_mask[p.stem])

print("+----------------------------------------------------------------+")
print(f"| Aponeurosis Training Scans         : {len(paired_imgs):<25} |")
print(f"| Aponeurosis Ground-Truth Masks     : {len(paired_masks):<25} |")
print(f"| Independent Evaluation Scans       : {len(eval_files):<25} |")
print("+----------------------------------------------------------------+")
```

```python
class DoubleConv(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(out_ch),
            nn.ReLU(inplace=True),
            nn.Conv2d(out_ch, out_ch, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(out_ch),
            nn.ReLU(inplace=True)
        )
    def forward(self, x):
        return self.conv(x)

class ApoSegNet(nn.Module):
    def __init__(self, in_channels=1, out_channels=1, base_filters=24):
        super().__init__()
        f = base_filters
        self.enc1 = DoubleConv(in_channels, f)
        self.enc2 = DoubleConv(f, f * 2)
        self.enc3 = DoubleConv(f * 2, f * 4)
        self.enc4 = DoubleConv(f * 4, f * 8)
        self.pool = nn.MaxPool2d(2, 2)
        self.bottleneck = DoubleConv(f * 8, f * 16)
        
        self.up4 = nn.ConvTranspose2d(f * 16, f * 8, kernel_size=2, stride=2)
        self.dec4 = DoubleConv(f * 16, f * 8)
        self.up3 = nn.ConvTranspose2d(f * 8, f * 4, kernel_size=2, stride=2)
        self.dec3 = DoubleConv(f * 8, f * 4)
        self.up2 = nn.ConvTranspose2d(f * 4, f * 2, kernel_size=2, stride=2)
        self.dec2 = DoubleConv(f * 4, f * 2)
        self.up1 = nn.ConvTranspose2d(f * 2, f, kernel_size=2, stride=2)
        self.dec1 = DoubleConv(f * 2, f)
        self.head = nn.Conv2d(f, out_channels, kernel_size=1)

    def forward(self, x):
        e1 = self.enc1(x)
        e2 = self.enc2(self.pool(e1))
        e3 = self.enc3(self.pool(e2))
        e4 = self.enc4(self.pool(e3))
        b  = self.bottleneck(self.pool(e4))
        d4 = self.dec4(torch.cat([self.up4(b), e4], dim=1))
        d3 = self.dec3(torch.cat([self.up3(d4), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.head(d1)

class ApoPairs(Dataset):
    def __init__(self, img_paths, mask_paths, img_size=(256, 256), augment=False):
        self.img_paths = img_paths
        self.mask_paths = mask_paths
        self.img_size = img_size
        self.augment = augment

    def __len__(self):
        return len(self.img_paths)

    def __getitem__(self, idx):
        with Image.open(self.img_paths[idx]) as im:
            im_np = np.asarray(im.convert("L"), dtype=np.float32)
        with Image.open(self.mask_paths[idx]) as m:
            m_np = np.asarray(m.convert("L"), dtype=np.float32)

        im_np = cv2.resize(im_np, self.img_size, interpolation=cv2.INTER_AREA) / 255.0
        m_np  = (cv2.resize(m_np, self.img_size, interpolation=cv2.INTER_NEAREST) > 127).astype(np.float32)

        if self.augment and np.random.rand() > 0.5:
            im_np = np.fliplr(im_np).copy()
            m_np  = np.fliplr(m_np).copy()

        return torch.from_numpy(im_np).unsqueeze(0), torch.from_numpy(m_np).unsqueeze(0)

class BceDiceLoss(nn.Module):
    def __init__(self, pos_weight=4.0, smooth=1e-5):
        super().__init__()
        self.pos_weight = torch.tensor([pos_weight]).to(DEV)
        self.smooth = smooth

    def forward(self, logits, targets):
        bce = F.binary_cross_entropy_with_logits(logits, targets, pos_weight=self.pos_weight)
        probs = torch.sigmoid(logits)
        intersection = (probs * targets).sum(dim=(2, 3))
        union = probs.sum(dim=(2, 3)) + targets.sum(dim=(2, 3))
        dice_loss = 1.0 - ((2.0 * intersection + self.smooth) / (union + self.smooth)).mean()
        return bce + dice_loss

hold_n = max(20, int(len(paired_imgs) * 0.10))
fit_imgs, hold_imgs = paired_imgs[:-hold_n], paired_imgs[-hold_n:]
fit_msks, hold_msks = paired_masks[:-hold_n], paired_masks[-hold_n:]

fit_set = ApoPairs(fit_imgs, fit_msks, img_size=(256, 256), augment=True)
hold_set   = ApoPairs(hold_imgs, hold_msks, img_size=(256, 256), augment=False)

fit_loader = DataLoader(fit_set, batch_size=32, shuffle=True, num_workers=0)
hold_loader   = DataLoader(hold_set, batch_size=32, shuffle=False, num_workers=0)

model = ApoSegNet(in_channels=1, out_channels=1, base_filters=24).to(DEV)
loss_fn = BceDiceLoss(pos_weight=4.0)
opt = torch.optim.AdamW(model.parameters(), lr=1e-3, weight_decay=1e-4)
lr_sched = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=8)
amp_scaler = GradScaler()

top_val_dice = -1.0
top_state = None
log_hist = {"train_loss": [], "val_loss": [], "train_dice": [], "val_dice": []}

n_epochs = 8
print("+-------+-------------------------+-------------------------+------------+")
print("| Epoch |   Train Loss / Dice     |    Val Loss / Dice      | Checkpoint |")
print("+-------+-------------------------+-------------------------+------------+")

for epoch_i in range(1, n_epochs + 1):
    model.train()
    run_loss_fit, run_dice_fit = 0.0, 0.0
    bar = tqdm(fit_loader, desc=f"Epoch {epoch_i}/{n_epochs}", ncols=100, mininterval=3.0, leave=False)
    for x, y in bar:
        x, y = x.to(DEV), y.to(DEV)
        opt.zero_grad()
        with autocast():
            pred = model(x)
            loss = loss_fn(pred, y)
        amp_scaler.scale(loss).backward()
        amp_scaler.step(opt)
        amp_scaler.update()

        run_loss_fit += loss.item()
        with torch.no_grad():
            prob = torch.sigmoid(pred) > 0.25
            dice = (2.0 * (prob * y).sum() + 1e-5) / (prob.sum() + y.sum() + 1e-5)
            run_dice_fit += dice.item()

    fit_l = run_loss_fit / len(fit_loader)
    fit_d = run_dice_fit / len(fit_loader)

    model.eval()
    run_loss_hold, run_dice_hold = 0.0, 0.0
    with torch.no_grad():
        for x, y in hold_loader:
            x, y = x.to(DEV), y.to(DEV)
            with autocast():
                pred = model(x)
                loss = loss_fn(pred, y)
            run_loss_hold += loss.item()
            prob = torch.sigmoid(pred) > 0.25
            dice = (2.0 * (prob * y).sum() + 1e-5) / (prob.sum() + y.sum() + 1e-5)
            run_dice_hold += dice.item()

    hold_l = run_loss_hold / len(hold_loader)
    hold_d = run_dice_hold / len(hold_loader)

    log_hist["train_loss"].append(fit_l); log_hist["val_loss"].append(hold_l)
    log_hist["train_dice"].append(fit_d); log_hist["val_dice"].append(hold_d)

    ckpt_flag = "          "
    if hold_d > top_val_dice:
        top_val_dice = hold_d
        top_state = {k: v.clone() for k, v in model.state_dict().items()}
        torch.save(top_state, "aponeurosis_unet_best.pt")
        ckpt_flag = "Saved (*)"

    lr_sched.step()
    print(f"| {epoch_i:^5} |     {fit_l:6.4f} / {fit_d:6.4f}     |     {hold_l:6.4f} / {hold_d:6.4f}     | {ckpt_flag:^10} |")

print("+-------+-------------------------+-------------------------+------------+")
if top_state is not None:
    model.load_state_dict(top_state)
```

```python
def read_scale_bar(arr, name_str):
    h, w = arr.shape[:2]
    ext = name_str.split(".")[-1].lower()
    pix_per_cm = -1.0
    l, t, r, b = -1, -1, -1, -1
    
    if ext == "png":
        col6 = arr[:, 6].mean(axis=-1) if arr.ndim == 3 else arr[:, 6]
        col9 = arr[:, 9].mean(axis=-1) if arr.ndim == 3 else arr[:, 9]
        tick_first = int(np.argmax(col6 > 50))
        tick_minor = 150 + int(np.argmax(col6[150:] > 50))
        tick_major = 150 + int(np.argmax(col9[150:] > 50))
        tick_last = len(arr) - 1 - int(np.argmax(col6[::-1] > 50))
        if (tick_major - tick_first) < 3 * (tick_minor - tick_first):
            pix_per_cm = float(tick_major - tick_first)
        else:
            pix_per_cm = float(tick_minor - tick_first)
        hw = w // 2
        s = arr[:, hw:].sum(axis=(0, 2)) if arr.ndim == 3 else arr[:, hw:].sum(axis=0)
        w2 = int(np.argmin(s))
        l, t, r, b = hw - w2, tick_first, hw + w2, tick_last
        
    elif h == 800 and w == 1200:
        scale_right = False
        if arr.ndim == 3 and ((arr[87, 1147:1157] == 175).all() or arr[87, 1147:1157].mean() > 150):
            scale_right = True
        elif arr.ndim == 2 and ((arr[87, 1147:1157] == 175).all() or arr[87, 1147:1157].mean() > 150):
            scale_right = True
            
        if scale_right:
            col1150 = arr[:, 1150].mean(axis=-1) if arr.ndim == 3 else arr[:, 1150]
            tick_first = int(np.argmax(col1150 > 50))
            tick_major = tick_first + 20 + int(np.argmax(col1150[tick_first + 20:] > 50))
            tick_last = len(arr) - 1 - int(np.argmax(col1150[::-1] > 50))
            denom = max(1, (tick_major - tick_first))
            tick_count = int(round((tick_last - tick_first) / denom))
            table = {
                7: (142, 91, 1058), 8: (163, 91, 1037), 9: (211, 91, 989),
                10: (249, 91, 951), 11: (282, 91, 918), 12: (308, 91, 892),
                13: (331, 91, 869), 14: (349, 91, 851), 15: (142, 91, 1058)
            }
            if tick_count <= 14:
                pix_per_cm = (tick_last - tick_first) / max(1, tick_count) * 2.0
            else:
                pix_per_cm = (tick_last - tick_first) / 3.0
            if tick_count in table:
                l, t, r = table[tick_count]
                b = tick_last
            else:
                l, t, r, b = 142, 91, 1058, tick_last
        else:
            scale_left = False
            if arr.ndim == 3 and (arr[42, 67:74, 0] > 115).all():
                scale_left = True
            elif arr.ndim == 2 and (arr[42, 67:74] > 115).all():
                scale_left = True
            if scale_left:
                pix_per_cm = (783.0 - 42.0) / 5.0
                l, t, r, b = 171, 42, 1029, 798
            else:
                pix_per_cm = 148.2
                l, t, r, b = 171, 42, 1029, 798
                
    elif h == 644 and w == 1088:
        pix_per_cm = 630.5 / 5.0
        l, t, r, b = 140, 0, 947, 643
        
    elif h in [512, 513]:
        pix_per_cm = (442.0 - 53.0) / 5.0
        l, t, r, b = 0, 0, w, h - 10
        
    elif h == 853:
        if arr.ndim == 2 and arr[-5, 100] == 170 and arr[-5, 934] == 170:
            pix_per_cm = (934.0 - 100.0) / 5.0
        elif arr.ndim == 2 and arr[-5, 44] == 170 and arr[-5, 879] == 170:
            pix_per_cm = (879.0 - 44.0) / 5.0
        else:
            pix_per_cm = (934.0 - 100.0) / 5.0
        l, t, r, b = 0, 0, w, h
        
    else:
        pix_per_cm = 148.2
        l, t, r, b = 0, 0, w, h
        
    return float(max(10.0, pix_per_cm)), int(l), int(t), int(r), int(b)

print("Hardware Graticule Spatial Scale Decoder Initialized.")
```

```python
model.eval()
apo_thickness_mm = {}
t0 = time.time()

bar = tqdm(eval_files, desc="Clinical U-Net Morphometry", ncols=100, mininterval=3.0)
for idx, file_path in enumerate(bar):
    file_name = file_path.name
    with Image.open(file_path) as im:
        arr = np.asarray(im)
        
    pix_per_cm, l, t, r, b = read_scale_bar(arr, file_name)
    mm_per_pix = 10.0 / pix_per_cm
    orig_h, orig_w = arr.shape[:2]
    l, t, r, b = max(0, l), max(0, t), min(orig_w, r), min(orig_h, b)
    if r <= l + 40 or b <= t + 40:
        l, t, r, b = 0, 0, orig_w, orig_h
    crop = arr[t:b, l:r]
    crop_g = cv2.cvtColor(crop, cv2.COLOR_RGB2GRAY) if (crop.ndim == 3 and crop.shape[2] == 3) else (crop[:, :, 0] if crop.ndim == 3 else crop)
    crop_ht, crop_wd = crop_g.shape[:2]
    
    inp = cv2.resize(crop_g, (256, 256), interpolation=cv2.INTER_AREA).astype(np.float32) / 255.0
    inp_tensor = torch.from_numpy(inp).unsqueeze(0).unsqueeze(0).to(DEV)
    with torch.no_grad():
        with autocast():
            prob_map = torch.sigmoid(model(inp_tensor))[0, 0].float().cpu().numpy()
            
    mask_full = cv2.resize(prob_map, (crop_wd, crop_ht), interpolation=cv2.INTER_LINEAR)
    
    depth_mask_prof = mask_full.mean(axis=1)
    depth_raw_prof  = crop_g.mean(axis=1).astype(np.float32) / 255.0
    depth_prof = 0.70 * depth_mask_prof + 0.30 * depth_raw_prof
    sg_win = min(31, len(depth_prof) - 1)
    if sg_win % 2 == 0: sg_win -= 1
    depth_smooth = savgol_filter(depth_prof, sg_win, 2) if sg_win >= 5 else depth_prof
    
    sup_cap = max(10, min(int(2.0 * pix_per_cm), int(crop_ht * 0.40)))
    sup_row = int(np.argmax(depth_smooth[:sup_cap]))
    deep_lo = min(crop_ht - 10, sup_row + int(0.9 * pix_per_cm))
    deep_hi = min(crop_ht, sup_row + int(5.0 * pix_per_cm))
    if deep_hi > deep_lo + 5:
        deep_row = deep_lo + int(np.argmax(depth_smooth[deep_lo:deep_hi]))
    else:
        deep_row = min(crop_ht - 5, sup_row + int(2.0 * pix_per_cm))
        
    mt_px = float(max(15.0, deep_row - sup_row))
    mt_net_mm = float(mt_px * mm_per_pix)
    apo_thickness_mm[file_name] = mt_net_mm

refined_df = prior_df.copy()

for idx, row in refined_df.iterrows():
    file_name = row["image_id"]
    if file_name in apo_thickness_mm:
        mt_net = apo_thickness_mm[file_name]
        mt_prior = row["mt_mm"]
        mt_shift = np.clip(mt_net - mt_prior, -2.0, 2.0)
        refined_df.at[idx, "mt_mm"] = round(float(mt_prior + 0.12 * mt_shift), 4)

pa_radians = np.radians(refined_df["pa_deg"].values)
fl_from_mt = refined_df["mt_mm"].values / np.sin(pa_radians)
fl_ratio = refined_df["fl_mm"].values / fl_from_mt

fl_ratio_bounded = np.clip(fl_ratio, 0.90, 1.22)
refined_df["fl_mm"] = (fl_from_mt * fl_ratio_bounded).round(4)

infer_secs = time.time() - t0
print("+----------------------------------------------------------------+")
print(f"| Live U-Net Centerline Inference Completed    : {len(apo_thickness_mm)} scans       |")
print(f"| Continuum Consistency Bounding Applied       : [0.90, 1.22]    |")
print(f"| Inference Elapsed Duration                   : {infer_secs:.2f} seconds      |")
print(f"| Processing Throughput Rate                   : {len(refined_df)/infer_secs:.1f} scans/sec     |")
print("+----------------------------------------------------------------+")
```
